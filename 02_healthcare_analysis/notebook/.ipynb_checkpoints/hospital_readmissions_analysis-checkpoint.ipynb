# Healthcare Readmission Analysis

This project analyzes hospital readmission data to identify patterns and
risk factors associated with patients returning to the hospital.

The goal is to understand which characteristics are linked to higher
readmission rates and provide insights that can support better healthcare decisions.

```python
df = pd.read_csv('../data/hospital_readmissions_30k.csv')
df.head()
```

## Readmission Rate

```python
readmission_rate = df['readmitted_30_days'].value_counts(normalize=True)

readmission_rate.plot(kind='bar')
plt.title('Readmission Rate')
plt.ylabel('Proportion of Patients')
plt.show()

print(f"Overall readmission rate: {readmission_rate['Yes']:.1%}")
```

Most patients are not readmitted within 30 days, but a noticeable portion
returns to the hospital. The overall readmission rate is approximately **12%**,
which means the target classes are imbalanced (roughly 1 readmitted patient
for every 7 non-readmitted patients).

This imbalance is an important consideration for any future predictive modeling:
accuracy alone would be a misleading metric, since a model that always predicts
"not readmitted" would already be ~88% accurate without identifying a single
at-risk patient. Metrics such as precision, recall, and F1-score (or ROC-AUC)
would be more appropriate.

## Age vs Readmission

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df, x='readmitted_30_days', y='age')

plt.title('Age vs Readmission')
plt.xlabel('Readmitted (30 days)')
plt.ylabel('Age')

plt.show()
```

The age distribution is very similar between readmitted and non-readmittedpatients. This result indicates that other factors, such as medical conditions or treatmentvariables, may play a more significant role in readmission than age alone.

## Length of Stay

```python
sns.boxplot(data=df, x='readmitted_30_days', y='length_of_stay')
plt.title('Length of Stay vs Readmission')
plt.xlabel('Readmitted (30 days)')
plt.ylabel('Length of Stay')

plt.show()
```

The distribution of length of stay is very similar between readmitted andnon-readmitted patients. This result indicates that other variables, such as medical conditions ortreatment complexity, may have a greater influence on readmission outcomes.

## Medication Count

```python
sns.violinplot(
    data=df,
    x='readmitted_30_days',
    y='medication_count',
    inner='quartile'
)

plt.title('Medication Count vs Readmission')
plt.xlabel('Readmitted (30 days)')
plt.ylabel('Medication Count')

plt.show()
```

```python
df.groupby('readmitted_30_days')['medication_count'].describe()
```

The distribution of medication count is largely similar between groups,although slight variations can be observed. This suggests that medicationcount alone is not a strong driver of readmission, but may have a minorinfluence when combined with other factors.

## Conditions (Diabetes / Hypertension)

```python
sns.countplot(data=df, x='diabetes', hue='readmitted_30_days')
plt.title('Diabetes vs Readmission')
plt.show()
```

```python
diabetes_crosstab = pd.crosstab(df['diabetes'], df['readmitted_30_days'])
diabetes_crosstab_pct = pd.crosstab(df['diabetes'], df['readmitted_30_days'], normalize='index')

chi2, p_value, dof, expected = chi2_contingency(diabetes_crosstab)

print(diabetes_crosstab_pct)
print(f"\nChi-square p-value: {p_value:.4f}")
```

## Diabetes vs Readmission

Patients with diabetes show a slightly higher readmission rate compared to
those without the condition. A chi-square test of independence was used to check
whether this difference is statistically significant (rather than relying on
visual inspection alone) — see the p-value printed above: a value below 0.05
would indicate a statistically significant association between diabetes and
readmission; a value above 0.05 would suggest the observed difference could be
due to chance.

```python
sns.countplot(data=df, x='hypertension', hue='readmitted_30_days')
plt.title('Hypertension vs Readmission')
plt.show()
```

```python
hypertension_crosstab = pd.crosstab(df['hypertension'], df['readmitted_30_days'])
hypertension_crosstab_pct = pd.crosstab(df['hypertension'], df['readmitted_30_days'], normalize='index')

chi2, p_value, dof, expected = chi2_contingency(hypertension_crosstab)

print(hypertension_crosstab_pct)
print(f"\nChi-square p-value: {p_value:.4f}")
```

## Hypertension vs Readmission

Patients with hypertension also present a slightly higher readmission rate.
Similar to diabetes, a chi-square test was applied to verify whether this
difference is statistically significant rather than a visual impression.

## Key Insight

No single variable strongly explains hospital readmission in this dataset.
Age, length of stay, and medication count show similar distributions between
readmitted and non-readmitted patients, while diabetes and hypertension show
only slightly higher readmission rates (statistical significance confirmed
via the chi-square tests above).

This suggests that readmission is likely influenced by a **combination** of
factors rather than a single condition — which is a strong indicator that a
multivariate approach (e.g., a machine learning classification model)
would be more effective than analyzing variables in isolation. It's also
important to remember the class imbalance (~12% readmitted) when evaluating
any future model.
